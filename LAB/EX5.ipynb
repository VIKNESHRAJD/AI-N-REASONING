# Bernoulli Dristribution


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p = 0.3
x = np.array([0,1])
probability = (p ** x) * ((1 - p) ** (1 - x))
print("Possible outcomes:", x)
print("Probababilities:", probability)
plt.bar(x,probability, alpha=0.7,label = "PMF")
plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.xticks(x, ["Failure (0)", "Success (1)"])
plt.legend()
plt.show()
```

# Binomial Distribution

```python
from scipy.stats import binom
```

```python
%%time
n = 8           
p = 0.5      
      
x = np.arange(n + 1)
probability = binom.pmf(x, n, p)
plt.bar(x, probability, alpha=0.5, label="PMF")
plt.plot(x, probability, marker="o", label="PMF line")
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.xticks(x)
plt.legend()
plt.grid(axis="y")
plt.show()
```

# Continuous Uniform Ditribution

```python
%time
from scipy.stats import uniform
x=uniform.rvs(loc=0.01,scale=0.99,size=100)
pdf_value=uniform.pdf(x[0],loc=0.01,scale=0.99)
print("PDF value=",pdf_value)
plt.hist(x,bins=10,density=True,alpha=0.7,label="Generated Data")
plt.axhline(y=1/0.99,color="black",linewidth=2,label="Theoretical PDF")
plt.title("Continous Uniform Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.show()
```
